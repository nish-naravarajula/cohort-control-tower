```python
import pandas as pd

A_DIR = "../data/tcga-brca/raw"
B_DIR = "../data/tcga-brca/releases/practice-B"

prot_A = pd.read_csv(f"{A_DIR}/protein_rppa.tsv.gz", sep="\t", index_col=0, dtype=str).T
prot_B = pd.read_csv(f"{B_DIR}/protein_rppa.tsv.gz", sep="\t", index_col=0, dtype=str).T

print(prot_A.shape, prot_B.shape)
```

```python
import hashlib

def hash_row(row):
    text = "|".join(row.values)
    return hashlib.sha256(text.encode()).hexdigest()

hash_A = prot_A.apply(hash_row, axis=1)
hash_B = prot_B.apply(hash_row, axis=1)

hash_A.head()
```
